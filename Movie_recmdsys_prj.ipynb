```python
# create a function convert()
def convert(obj):
    L = []
    for i in ast.literal_eval(obj):
        L.append(i['name'])
    return L
```

```python
# we have a module 
import ast
ast.literal_eval('[{"id": 28, "name": "Action"}, {"id": 12, "name": "Adventure"}, {"id": 14, "name": "Fantasy"}, {"id": 878, "name": "Science Fiction"}]')
```

```python
movie['genres'] = movie['genres'].apply(convert)
```

```python
movie.head(2)
# we resoleve the genere column will do same thing with keywords
```

```python
movie['keywords'] = movie['keywords'].apply(convert)
```

```python
movie.head(2)
```

```python
#now we goonna format the cast 
# we reqire first 3 dict
def convert3(obj):
    L = []
    counter = 0
    for i in ast.literal_eval(obj):
        if counter != 3:
            L.append(i['name'])
            counter += 1
        else:
            break
    return L
```

```python
movie['cast'] = movie['cast'].apply(convert3)
```

```python
movie.head(2)
```

```python
# now for crew we only need dict which means JOB = director
```

```python
def fetch_director(obj):
    L = []
    counter = 0
    for i in ast.literal_eval(obj):
        if i['job'] == 'Director':
            L.append(i['name'])
            counter += 1
            break
    return L
```

```python
movie['crew'] = movie['crew'].apply(fetch_director)
movie.head(2)
```

```python
# Now overview col is a string so we'ill covert it in a list 
```

```python
movie['overview'] = movie['overview'].apply(lambda x:x.split())
```

```python
movie.head(2)
```

```python
# Now we  will do a transformation in cols i.e we'll remove the space between the word 
# Like 'Sam Worthington'  to "SamWorthington"
```

```python
movie['genres'] = movie['genres'].apply(lambda x:[i.replace(" ","") for i in x])
movie['keywords'] = movie['keywords'].apply(lambda x:[i.replace(" ","") for i in x])
movie['cast'] = movie['cast'].apply(lambda x:[i.replace(" ","") for i in x])
movie['crew'] = movie['crew'].apply(lambda x:[i.replace(" ","") for i in x])
```

```python
movie.head(2)
```

```python
# Now we'll create a new col and concatinate all the cols in tag col
movie['tags'] = movie['overview'] + movie['genres'] + movie['keywords'] + movie['cast'] + movie['crew']
```

```python
from sklearn.feature_extraction.text import CountVectorizer
cv = CountVectorizer(max_features=5000,stop_words='english')
```

```python
vector = cv.fit_transform(new_df['tags']).toarray()
```

```python
# now we'll apply stemming to remoove simaliar word like [dance,danced,dancing] to [dance,dance,dance]
# !pip install nltk
```

```python
import nltk
```

```python
from nltk.stem.porter import PorterStemmer
ps = PorterStemmer()
```

```python
def stem(text):
    y = []
    for i in text.split():
        y.append(ps.stem(i))

    return " ".join(y)
        
```

```python
stem('In the 22nd century, a paraplegic Marine is dispatched to the moon Pandora on a unique mission, but becomes torn between following orders and protecting an alien civilization. Action Adventure Fantasy ScienceFiction cultureclash future spacewar spacecolony society spacetravel futuristic romance space alien tribe alienplanet cgi marine soldier battle loveaffair antiwar powerrelations mindandsoul 3d SamWorthington ZoeSaldana SigourneyWeaver JamesCameron')
```

```python
new_df['tags'] = new_df['tags'].apply(stem)
```

```python
def recommend(movie):
    movie_index = new_df[new_df['title'] == movie].index[0]
    distances = similarity[movie_index]
    movie_list = sorted(list(enumerate(distances)),reverse = True,key=lambda x:x[1])[1:6]

    for i in movie_list:
        print(new_df.iloc[i[0]].title)
```
